```python
import os
from pathlib import Path
```

```python
def setup_project_folders(org_path: str):
    """
    미니 프로젝트용 기본 디렉토리 생성:
      - data/raw
      - data/processed
      - data/predictions
      - data/models
      - notebooks
      - outputs/eda
      - outputs/logs
    """

    base = Path(org_path)

    folders = [
        base / "data/raw",           # 원본 데이터
        base / "data/processed",     # 전처리/학습용 데이터
        base / "data/predictions",   # 예측 결과 저장
        base / "data/models",        # 학습된 모델 저장

        base / "notebooks",          # 노트북 정리 공간

        base / "outputs/eda",        # 탐색 분석 결과
        base / "outputs/logs",       # 학습·실험 로그
    ]

    for folder in folders:
        folder.mkdir(parents=True, exist_ok=True)
        print(f"[✔] 생성됨: {folder}")

    print("\n모든 폴더가 준비되었습니다.")

# 예시 사용
org_path = "C:/Users/Owner/Desktop/지윤/3.기타/프로젝트/purchase-conversion-prediction/"
setup_project_folders(org_path)
```
